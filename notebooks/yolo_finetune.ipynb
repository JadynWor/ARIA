combine data set HERIDAL & SARD

```python
import os, shutil, yaml

for split in ['train', 'valid']:
    for sub in ['images', 'labels']:
        os.makedirs(f'/content/drive/MyDrive/ARIA/data/combined/{split}/{sub}', exist_ok=True)

for dataset in ['heridal', 'sard']:
    for split in ['train', 'valid']:
        for sub in ['images', 'labels']:
            src = f'/content/drive/MyDrive/ARIA/data/{dataset}/{split}/{sub}'
            dst = f'/content/drive/MyDrive/ARIA/data/combined/{split}/{sub}'
            for f in os.listdir(src):
                shutil.copy2(os.path.join(src, f), os.path.join(dst, f'{dataset}_{f}'))

print("Done copying!")
```

```python
data_config = {
    'train': '/content/drive/MyDrive/ARIA/data/combined/train/images',
    'val': '/content/drive/MyDrive/ARIA/data/combined/valid/images',
    'nc': 1,
    'names': ['human']
}

with open('/content/drive/MyDrive/ARIA/data/combined/data.yaml', 'w') as f:
    yaml.dump(data_config, f)

print("Config created!")
```

```python
from ultralytics import YOLO

model = YOLO('/content/drive/MyDrive/ARIA/runs/heridal-finetune-v2/weights/best.pt')

results = model.train(
    data='/content/drive/MyDrive/ARIA/data/combined/data.yaml',
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    project='/content/drive/MyDrive/ARIA/runs',
    name='combined-finetune'
)
```
